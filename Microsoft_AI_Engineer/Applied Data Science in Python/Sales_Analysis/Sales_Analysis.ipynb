# AAL Sales Analysis Q4 2020

This project analyzes the fourth-quarter sales data for AAL across different states, demographic groups, and times of day. The objective is to identify high-performing regions and demographic segments while providing insights to assist the Sales and Marketing team in making data-driven decisions.

## Data Loading

The required libraries are imported and the dataset containing fourth-quarter apparel sales in Australia is loaded into a pandas DataFrame for analysis.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("AusApparalSales4thQrt2020.csv")

```

## Data Wrangling

The dataset is inspected to understand its structure, identify missing values, and verify that the data is suitable for further analysis. The functions `head()`, `info()`, and `isna()` are used to explore the dataset and check for null or incorrect entries.

```python
df.head()
df.info()
df.isna().sum()
```

## Data Cleaning

To ensure the original dataset remains unchanged during preprocessing, a copy of the dataset is created. This cleaned dataset will be used for all subsequent transformations and analysis.
The dataset contained no missing values, so no rows were dropped and no imputation was required.

```python
df_clean = df.copy()

df_clean.head()
```

## Data Normalization

Normalization was applied using MinMaxScaler to scale the Unit and Sales values between 0 and 1. This ensures that large values do not dominate the analysis and allows fair comparison across states and demographic groups.

```python

scaler = MinMaxScaler()

df_clean[["Unit_Normalized", "Sales_Normalized"]] = scaler.fit_transform(df_clean[["Unit", "Sales"]])

df_clean[["Unit_Normalized", "Sales_Normalized"]].describe()
```

## Data Analysis

Descriptive statistical analysis is performed on the Sales and Unit columns. The metrics calculated include the mean, median, mode, and standard deviation, which provide insight into the distribution and variability of the sales data.

```python
df_clean["Sales"].mean()
df_clean["Sales"].median()
df_clean["Sales"].mode()   
df_clean["Sales"].std()

df_clean["Unit"].mean()
df_clean["Unit"].median()
df_clean["Unit"].mode()
df_clean["Unit"].std()

print("Sales - Mean:", df_clean["Sales"].mean())
print("Sales - Median:", df_clean["Sales"].median())
print("Sales - Mode:", df_clean["Sales"].mode()[0])
print("Sales - Standard Deviation:", df_clean["Sales"].std())

print("Unit - Mean:", df_clean["Unit"].mean())
print("Unit - Median:", df_clean["Unit"].median())
print("Unit - Mode:", df_clean["Unit"].mode()[0])
print("Unit - Standard Deviation:", df_clean["Unit"].std())
```

## Sales Reports

Weekly, monthly, and quarterly reports are generated to summarize sales trends over time. These reports provide insights into how sales vary throughout the quarter and help identify periods of higher or lower activity.
The groupby() function is appropriate for data chunking in this project because it allows sales to be aggregated by state, group, week, month, quarter, and time of day. This makes it easier to compare business performance across categories.

```python
df_clean[["Unit", "Sales"]].corr()
df_clean.groupby("State")[["Unit", "Sales"]].sum()
df_clean.groupby("Time")[["Unit", "Sales"]].sum()
df_clean.groupby("Group")[["Unit", "Sales"]].sum().sort_values(by="Sales", ascending=False)
group_sales = df_clean.groupby("Group")[["Unit", "Sales"]].sum().sort_values(by="Sales", ascending=False)
state_sales = df_clean.groupby("State")[["Unit", "Sales"]].sum().sort_values(by="Sales", ascending=False)

print("Highest Sales Group:\n", group_sales.head(1).reset_index().to_string(index=False),
      "\n\nLowest Sales Group:\n", group_sales.tail(1).reset_index().to_string(index=False),
      "\n\nSales by State:\n", state_sales,
      "\n\nHighest Sales State:\n", state_sales.head(1).reset_index().to_string(index=False),
      "\n\nLowest Sales State:\n", state_sales.tail(1).reset_index().to_string(index=False))
```

```python
df_clean["Date"] = pd.to_datetime(df_clean["Date"],)
df_clean["Week"] = df_clean["Date"].dt.isocalendar().week
df_clean["Month"] = df_clean["Date"].dt.month
df_clean["Quarter"] = df_clean["Date"].dt.quarter
daily_sales = df_clean.groupby("Date")[["Unit", "Sales"]].sum()
weekly_sales = df_clean.groupby("Week")[["Unit", "Sales"]].sum()
monthly_sales = df_clean.groupby("Month")[["Unit", "Sales"]].sum()
quarterly_sales = df_clean.groupby("Quarter")[["Unit", "Sales"]].sum()
time_sales = df_clean.groupby("Time")[["Sales"]].sum()

print("Daily Sales:\n\n", daily_sales.reset_index().to_string(index=False),
    "\n\nWeekly Sales:\n\n", weekly_sales.reset_index().to_string(index=False),
    "\n\nMonthly Sales:\n\n", monthly_sales.reset_index().to_string(index=False),
    "\n\nQuarterly Sales:\n\n", quarterly_sales.reset_index().to_string(index=False))
```

## Data Visualization

```python
plt.figure()
sns.barplot(data=state_sales.reset_index(), x="State", y="Sales")
plt.title("Total Sales by State")
plt.xlabel("State")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()

plt.figure()
sns.barplot(data=group_sales.reset_index(), x="Group", y="Sales")
plt.title("Total Sales by Group")
plt.show()

plt.figure()
sns.barplot(data=time_sales.reset_index(), x="Time", y="Sales")
plt.title("Total Sales by Time of Day")
plt.show()

plt.figure()
sns.barplot(data=daily_sales.reset_index(), x="Date", y="Sales")
plt.title("Total Sales by Day")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.show()

plt.figure()
sns.boxplot(data=df_clean[["Sales", "Unit"]])
plt.title("Box Plot of Sales and Units")
plt.xlabel("Sales")
plt.ylabel("Units")
plt.show()

plt.figure()
sns.boxplot(data=df_clean, x="Group", y="Sales")
plt.title("Sales Distribution by Group")
plt.show()

plt.figure()
sns.displot(df_clean["Sales"])
plt.title("Distribution of Sales")
plt.show()
```

## Insights and Recommendations

Based on the analysis conducted:

• Certain states generate significantly higher sales revenue compared to others.

• Sales vary across demographic groups, with some groups contributing more to overall revenue.

• Sales activity differs by time of day, indicating potential peak purchasing periods.

### Recommendations

• Expand marketing campaigns in lower-performing states.

• Focus promotions and inventory planning on the highest-performing demographic groups.

• Use peak sales periods to deploy targeted promotions and hyper-personalized offers.
